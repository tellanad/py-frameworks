```python
import os
from google.colab import userdata
from langchain_core.tools import tool          # Native LangChain tool wrapper
from langgraph.prebuilt import create_react_agent
from langchain_groq import ChatGroq

# 1. Setup authentication for Groq
os.environ["GROQ_API_KEY"] = userdata.get('GROQ_API_KEY')
model = ChatGroq(model="openai/gpt-oss-120b")

# 2. Define your tool using native LangChain decorators (No network/MCP required)
@tool
def search_langchain_memory_docs(query: str) -> str:
    """Searches LangChain documentation specifically regarding adding short-term memory layers."""
    return (
        "According to the LangChain documentation, short-term memory can be added to an agent "
        "by wrapping it with a state checker or using LangGraph's MemorySaver checkpointer. "
        "You initialize a MemorySaver() and pass it into the compile step of a graph: "
        "agent = workflow.compile(checkpointer=MemorySaver())."
    )

# 3. Consolidate your tools into a standard list
tools = [search_langchain_memory_docs]

# 4. Construct the functional agent structure
agent = create_react_agent(model, tools)

# 5. Execute using notebook top-level await syntax
response = await agent.ainvoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "How do I add short-term memory to a LangChain agent?",
            }
        ]
    }
)

# 6. Output the content response
print(response["messages"][-1].content)
```
